```python
#1. Input DNA Sequence
from Bio.Seq import Seq

dna = Seq("ATGGCTTTTTAA")

print("DNA Sequence:", dna)
```

```python
#2. Transcription Process
mrna = dna.transcribe()

print("mRNA Sequence:", mrna)
```

```python
#3. Translation Process
protein = mrna.translate()

print("Protein Sequence:", protein)
```

Keterangan:
- M = Methionine
- A = Alanine
- F = Phenylalanine
- *= Stop codon

```python
#4. Identify Start and Stop Codon
start_codon = "AUG"
stop_codons = ["UAA", "UAG", "UGA"]

print("Start Codon:", start_codon)

for codon in stop_codons:
    if codon in mrna:
        print("Stop Codon Found:", codon)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 4))

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis("off")

# Title
ax.set_title(
    "Central Dogma of Molecular Biology",
    fontsize=16,
    pad=20
)

# DNA box
ax.text(
    0.15, 0.5,
    "DNA\nATG GCT TTT TAA",
    ha="center",
    va="center",
    fontsize=12,
    bbox=dict(
        boxstyle="round,pad=0.5",
        facecolor="lightblue",
        edgecolor="black"
    )
)

# mRNA box
ax.text(
    0.5, 0.5,
    "mRNA\nAUG GCU UUU UAA",
    ha="center",
    va="center",
    fontsize=12,
    bbox=dict(
        boxstyle="round,pad=0.5",
        facecolor="lightgreen",
        edgecolor="black"
    )
)

# Protein box
ax.text(
    0.85, 0.5,
    "Protein\nMet-Ala-Phe",
    ha="center",
    va="center",
    fontsize=12,
    bbox=dict(
        boxstyle="round,pad=0.5",
        facecolor="plum",
        edgecolor="black"
    )
)

# Transcription arrow
ax.annotate(
    "",
    xy=(0.39, 0.5),
    xytext=(0.27, 0.5),
    arrowprops=dict(
        arrowstyle="->",
        lw=1.5,
        color="black"
    )
)

# Transcription label
ax.text(
    0.33, 0.63,
    "Transcription",
    ha="center",
    va="center",
    fontsize=11
)

# Translation arrow
ax.annotate(
    "",
    xy=(0.74, 0.5),
    xytext=(0.61, 0.5),
    arrowprops=dict(
        arrowstyle="->",
        lw=1.5,
        color="black"
    )
)

# Translation label
ax.text(
    0.675, 0.63,
    "Translation",
    ha="center",
    va="center",
    fontsize=11
)

plt.tight_layout()
plt.show()
```
